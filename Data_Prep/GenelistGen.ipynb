```python
import pandas as pd
import numpy as np
```

```python
import os
os.sys.path.insert(0 , '../../MOGDx/Modules/PNetTorch/MAIN/')
from utils import *

gmt = GMT()
df = gmt.load_data('./../../MOGDx/Modules/PNetTorch/reactome/ReactomePathways.gmt', pathway_col=1, genes_col=3)

all_genes = df['gene'].unique()
```

```python
pd_genelist = pd.read_csv('PD_Genelists.csv', header=0)
```

```python
for i , gen in enumerate(pd_genelist.columns) : 
    pos_df = pd.DataFrame((set(all_genes) & set(pd_genelist[gen])) , columns=['genes'])
    pos_df['group'] = ['pos' for i in range(pos_df.shape[0])]
    
    neg_df = pd.DataFrame((set(all_genes) - set(pd_genelist[gen])) , columns=['genes'])
    neg_df['group'] = ['neg' for i in range(neg_df.shape[0])]

    print( pd.concat([pos_df , neg_df.sample(pos_df.shape[0]*3)])['group'].value_counts())
    
    pd.concat([pos_df , neg_df.sample(pos_df.shape[0]*3)]).to_csv(f'./../../data/PPMI/03_final/ext_data/genelist_gen{i}.txt', sep='\t', index=False)
```
