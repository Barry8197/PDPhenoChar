# Data Preparation

The goal will be to obtain a sample for every idiopathic patient we have a subgroup assigned to. In each modality we will take the latest available time point sample.

```python
import pandas as pd
import pickle
```

```python
patnos = pd.read_csv('./../idiopathic_subgroup.csv' , index_col=0)['PATNO'].astype(str)
```

```python
events = ['BL' , 'V02' , 'V04' , 'V06' , 'V08' , 'V10' , 'V12' , 'V14' , 'V16']
```

### mRNA

```python
mod = 'mRNA'
datExpr = pd.DataFrame()
for event in reversed(sorted(events)) : 
    print(event)
    try : 
        with open(f'./../../data/PPMI/02_working/{event}/{mod}_preprocessed.pkl' , 'rb') as file : 
            loaded_data = pickle.load(file)

        datExpr_tmp = loaded_data['datExpr']
        new_patnos = set(datExpr_tmp.index) - set(datExpr.index)
        
        overlap = list(new_patnos & set(patnos))
        datExpr_tmp = datExpr_tmp.loc[overlap]

        datExpr = pd.concat([datExpr , datExpr_tmp])
    except : 
        pass

with open(f'./../../data/PPMI/02_working/GIMR/{mod}_preprocessed.pkl' , 'wb') as file : 
    pickle.dump({'datExpr' : datExpr} , file)
```
